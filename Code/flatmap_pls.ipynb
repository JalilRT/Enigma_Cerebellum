## **PLS - Full sample**

### MDTB

```python
# 1. Load your label GIFTI file
MDTB_label_gii_path = 'utils/suit/flatmap/MDTB_10Regions.label.gii'

# 2. Load custom colormap (skip header)
MDTB_lv8 = "Figures/Flatmap/MDTB/flatmap/LV_8_PLS_MDTB_rgb4Flatmap.txt"
MDTB_lv9 = "Figures/Flatmap/MDTB/flatmap/LV_9_PLS_MDTB_rgb4Flatmap.txt"
MDTB_lv10 = "Figures/Flatmap/MDTB/flatmap/LV_10_PLS_MDTB_rgb4Flatmap.txt"

dir_path = "Figures/Flatmap/MDTB/flatmap_maps"
os.makedirs(dir_path, exist_ok=True)
print(f"Directory ready: {dir_path}")

for level, cmap_file in [('LV8', MDTB_lv8), ('LV9', MDTB_lv9), ('LV10', MDTB_lv10)]:
    rgb_colors = np.loadtxt(cmap_file, skiprows=1)
    background = np.array([[0.95, 0.95, 0.95]])  # or white [1,1,1]
    rgb_full = np.vstack([background, rgb_colors])
    cmap = ListedColormap(rgb_full)

    ax = flatmap.plot(
        data=MDTB_label_gii_path,
        cmap=cmap,
        bordersize=0.5,
        render='matplotlib',
        new_figure=True
    )
    plt.gca().set_facecolor('none')
    plt.axis('off')
    ax.get_figure().savefig(
        os.path.join(dir_path, f"MDTB_{level}.png"),
        dpi=300,
        transparent=True,
        bbox_inches='tight',
        pad_inches=0
    )
    plt.close(ax.get_figure())
```
