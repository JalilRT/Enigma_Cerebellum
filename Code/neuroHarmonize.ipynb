## Run the harmonizer

```python
data = pd.read_csv(os.path.join(path, "Data/Harmonizer/","Volumes_Cerebellum_raw_lessNA_Acapulco.csv"))
NI_features = data.columns.tolist()[3::]
covars_features = data.columns.tolist()[0:3]
```

```python
print('# --------------------------------- harmonizer --------------------------------- #')
### Harmonize the entire dataset ###
print('### Harmonize the entire dataset ###')
harm = harmonizer(feature_names = NI_features, covariates_names = covars_features, eb = True, smooth_terms = ["Age"], smooth_term_bounds=(0, 100))
data_adj = harm.fit_transform(data)
print()
```

```python
# load your data and all numeric covariates
Volumes_data = pd.read_csv('Data/RData/Volumes_raw.csv')
Volumes = Volumes_data[["CSF", "GM", "WM", "ICV"]]
Volumes_array = np.array(Volumes)
covars = Volumes_data[["Sex", "Age", "Scanner"]].rename(columns={"Scanner": "SITE"})
covars['Sex'] = covars['Sex'].map({'F': 0, 'M': 1})
```

```python
# run harmonization with NONLINEAR effects of age
my_model, my_data_adj = harmonizationLearn(Volumes_array, covars, smooth_terms=['Age'])
```

```python
my_data_adj_df = pd.DataFrame(my_data_adj, columns=Volumes.columns)
my_data_adj_df.to_csv('Data/RData/Volumes_neuroHarmonized.csv', index=False)
```
