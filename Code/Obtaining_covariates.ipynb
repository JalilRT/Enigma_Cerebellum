## **8 Obtaining Covariables**

### **8.1 Data preparation**

```python
path="/media/jalil/ADATA HD710 PRO/Enigma_VBM/VBM"
lv_path=paste0(path,"/")
DB_path=paste0(path,"/wdseg1_files")
DB_path=paste0(path,"/MRI")
Cov_path=paste0(path,"/Covariables")
```

```python
# Reading volumes and demographics file

# Reading filenames from SUD-DB paths
SUD_names <- list.dirs(DB_path,full.names = F) %>% tail(n = -1)
SUD_filenames <- 1:length(SUD_names) %>% 
  map(~ list.files(paste0(DB_path,"/",SUD_names[.x])) %>% as_tibble()) %>% 
  set_names(SUD_names)

# Removing quotes and extensions
SUD_filenames_clean <- SUD_filenames %>% 
  map(~ .x$value %>% str_remove("wd") %>% str_remove("_n4_mni_seg1.nii") )

SUD_fn_tibble <- SUD_filenames %>% 
  map(~ .x %>% mutate(ID_MRI = value %>% 
                        str_remove("wd") %>% 
                        str_remove("_n4_mni_seg1.nii")) %>% 
        rename("ID_long" = "value") ) %>% reduce(rbind)
```

### **8.1 Data joining**

```python
SUD_VBM_covs <- SUD_fn_tibble %>% unique() %>% full_join(Volumes_Cerebellum %>% 
                              select(ID_MRI,Substance,Age,Sex,Control,Total_Cerebel_Vol,Site), 
                            by = "ID_MRI") %>% drop_na()
```

```python
list_VBM_covs <- split(SUD_VBM_covs, f = SUD_VBM_covs$Substance)
```

```python
names(list_VBM_covs) %>% map(~ list_VBM_covs[[.x]] %>% select(ID_MRI) %>% as.vector() %>%
                             write.table(file = paste0(Cov_path,"/ID/",.x,".txt"),col.names = F,row.names = F,quote=FALSE) )
```

```python
SUD_VBM_covs$Age %>% as.vector() %>% round() %>%
    write.table(file = paste0(Cov_path,"/Age/","Age.txt"),col.names = F,row.names = F,quote=FALSE)
```
