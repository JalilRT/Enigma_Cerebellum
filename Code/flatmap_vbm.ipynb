```python
dir_path = "Figures/Flatmap/VBM/full"
os.makedirs(dir_path, exist_ok=True)
print(f"Directory ready: {dir_path}")

for level, data_file in [('users', User_HC), ('ALC', ALC_HC), ('ATS', ATS_HC), ('CANN', CANN_HC), ('COC', COC_HC), ('NIC', NIC_HC)]:  
    # Define a custom undermap with the color #E7E7E7
    undermap_color = "#E7E7E7"
    custom_undermap = ListedColormap([undermap_color])

    ax = flatmap.plot(data=data_file, cmap=cmap, \
        cscale = [-8,7], \
        threshold=[-1.959,1.959], \
        new_figure=True, \
        undermap=custom_undermap, \
        underscale=None, \
        bordersize = 0.5, \
        render='matplotlib')

    plt.gca().set_facecolor('none')
    plt.axis('off')
    ax.get_figure().savefig(
        os.path.join(dir_path, f"VBM_{level}.png"),
        dpi=300,
        transparent=True,
        bbox_inches='tight',
        pad_inches=0
    )
    plt.close(ax.get_figure())
```
