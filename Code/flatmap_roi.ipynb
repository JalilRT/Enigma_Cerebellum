## **T-values - Matchit sample**

### MDTB

```python
# 1. Load your label GIFTI file
MDTB_label_gii_path = 'utils/suit/flatmap/MDTB_10Regions.label.gii'

# 2. Load custom colormap (skip header)
MDTB_users = "Figures/Flatmap/MDTB/permutation_flatmap/users_perm_MDTB_rgb4Flatmap_tval.txt"
MDTB_ALC = "Figures/Flatmap/MDTB/permutation_flatmap/ALC_perm_MDTB_rgb4Flatmap_tval.txt"
MDTB_ATS = "Figures/Flatmap/MDTB/permutation_flatmap/ATS_perm_MDTB_rgb4Flatmap_tval.txt"
MDTB_CANN = "Figures/Flatmap/MDTB/permutation_flatmap/CANN_perm_MDTB_rgb4Flatmap_tval.txt"
MDTB_COC = "Figures/Flatmap/MDTB/permutation_flatmap/COC_perm_MDTB_rgb4Flatmap_tval.txt"
MDTB_NIC = "Figures/Flatmap/MDTB/permutation_flatmap/NIC_perm_MDTB_rgb4Flatmap_tval.txt"
MDTB_OPI = "Figures/Flatmap/MDTB/permutation_flatmap/OPI_perm_MDTB_rgb4Flatmap_tval.txt"

dir_path = "Figures/Flatmap/MDTB/permutation_flatmap_maps"
os.makedirs(dir_path, exist_ok=True)
print(f"Directory ready: {dir_path}")

for level, cmap_file in [('users', MDTB_users), ('ALC', MDTB_ALC), ('ATS', MDTB_ATS), ('CANN', MDTB_CANN), ('COC', MDTB_COC), ('NIC', MDTB_NIC), ('OPI', MDTB_OPI)]:
    rgb_colors = np.loadtxt(cmap_file, skiprows=1)
    background = np.array([[0.95, 0.95, 0.95]])  # or white [1,1,1]
    rgb_full = np.vstack([background, rgb_colors])
    cmap = ListedColormap(rgb_full)

    ax = flatmap.plot(
        data=MDTB_label_gii_path,
        cmap=cmap,
        bordersize=0.5,
        render='matplotlib',
        new_figure=True
    )
    plt.gca().set_facecolor('none')
    plt.axis('off')
    ax.get_figure().savefig(
        os.path.join(dir_path, f"tvalues_MDTB_{level}.png"),
        dpi=300,
        transparent=True,
        bbox_inches='tight',
        pad_inches=0
    )
    plt.close(ax.get_figure())
```
